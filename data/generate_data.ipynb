```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
## APPL STOCKS DATA

data = yf.download(
    "AAPL",
    start="2015-01-01",
    end="2025-01-01",
    interval="1d",
    auto_adjust=True
)

monthly_avg = (
    data['Close']['AAPL']
    .resample('ME')
    .mean()
)

numpy_data = monthly_avg.reset_index()['AAPL'].to_numpy()
np.save('aapl_stock_monthly.npy', numpy_data)

plt.plot(numpy_data)
```

```python
## Delhi Tempreture Data

df_train = pd.read_csv('DailyDelhiClimateTrain.csv')
df_test = pd.read_csv('DailyDelhiClimateTest.csv')
df_train = pd.concat([df_train, df_test])
df = df_train.reset_index()['meantemp'].to_numpy()
np.save('daily_temp_delhi.npy', df)
```

```python
import pandas as pd
df = pd.read_csv('indian_nifty50_stocks_10y_daily.csv')
```

```python
stocks = df['Ticker'].unique()
stocks
```

```python
import numpy as np
```

```python
stock =  'TRENT.NS'
stock_name = stock.split('.')[0].lower()
stock_name
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 7))

for stock in stocks:
    stock_name = stock.split('.')[0].lower()

    df_nifty = df[df['Ticker'] == stock].dropna().reset_index()
    nifty = df_nifty['Close'].to_numpy()

    # np.save(f'{stock_name}_l10y.npy', nifty)

    plt.plot(nifty, label=stock_name.upper())

plt.xlabel('Time')
plt.ylabel('Close Price')
plt.title('Stock Price Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
